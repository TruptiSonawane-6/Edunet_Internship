```python
import numpy as np
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
```

```python
x_trained=data['actual']
y_pred=data['predicted']
```

```python

conf_matrix = confusion_matrix(x_trained, y_pred)
accuracy = accuracy_score(x_trained, y_pred)
precision = precision_score(x_trained, y_pred)
recall = recall_score(x_trained, y_pred)
f1 = f1_score(x_trained, y_pred)

print("Confusion Matrix:\n", conf_matrix)
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
```
