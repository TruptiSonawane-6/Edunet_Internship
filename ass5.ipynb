```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.linear_model import LinearRegression

df = pd.read_csv("C:/Users/tsson/Downloads/drive-download-20251102T105633Z-1-001/Assignment 5/temperatures.csv")
print("columns")
print(df.columns)
print("5 rows")
df.head()
```

```python
mar_may_avg = df['MAR-MAY']
print(mar_may_avg)
```

```python
df_mar_may = df[['MAR-MAY', 'ANNUAL']].dropna()
print(df_mar_may.head())
```

```python
x = df_mar_may[['MAR-MAY']]
y = df_mar_may['ANNUAL']

y
```

```python
model = LinearRegression()
model.fit(x, y)

y_pred = model.predict(x)

mse = mean_squared_error(y, y_pred)
mae = mean_absolute_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("mse:", mse)
print("mae:", mae)
print("R^2:", r2)
```

```python
plt.figure(figsize=(10, 5))
plt.scatter(x, y, color='green', label='Actual Annual Avg Temp')
plt.plot(x, y_pred, color='orange', label='Regression Line')
plt.xlabel('MAR-MAY Avg Temperature(*c)')
plt.ylabel('Annual avg Temp(*C)')
plt.title('MAR-MAY vs Annual Avg Temperature')
plt.legend()
plt.grid(True)
plt.show()
```
